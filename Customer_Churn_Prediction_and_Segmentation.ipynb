```python
# ================================================================
# INTELLIGENT CUSTOMER CHURN RISK PREDICTION &
# BEHAVIOURAL SEGMENTATION SYSTEM
#
# COMPLETE GOOGLE COLAB IMPLEMENTATION
# ================================================================

# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

import os
import glob
import shutil
import warnings

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split, GridSearchCV

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    roc_curve,
    silhouette_score
)

from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

import joblib

print("=" * 75)
print("CUSTOMER CHURN PREDICTION & BEHAVIOURAL SEGMENTATION SYSTEM")
print("=" * 75)
print("\nAll libraries imported successfully.")


# ================================================================
# 2. CREATE OUTPUT DIRECTORIES
# ================================================================

BASE_DIR = "/content/customer_churn_project"

FIGURE_DIR = os.path.join(BASE_DIR, "figures")
MODEL_DIR = os.path.join(BASE_DIR, "models")
RESULT_DIR = os.path.join(BASE_DIR, "results")

os.makedirs(FIGURE_DIR, exist_ok=True)
os.makedirs(MODEL_DIR, exist_ok=True)
os.makedirs(RESULT_DIR, exist_ok=True)

print("\nOutput directories created.")


# ================================================================
# 3. UPLOAD DATASET
# ================================================================

print("\n" + "=" * 75)
print("UPLOAD KAGGLE DATASET")
print("=" * 75)

print("""
Please upload:
WA_Fn-UseC_-Telco-Customer-Churn.csv
""")

uploaded = files.upload()

csv_files = glob.glob("/content/*.csv")

if len(csv_files) == 0:
    raise FileNotFoundError(
        "No CSV file was uploaded. Please upload the Telco Customer Churn CSV."
    )

# Prefer the expected dataset if multiple CSVs exist
expected_files = [
    f for f in csv_files
    if "Telco" in os.path.basename(f)
    or "telco" in os.path.basename(f)
]

if expected_files:
    file_path = expected_files[0]
else:
    file_path = csv_files[0]

df = pd.read_csv(file_path)

print("\nDataset loaded successfully.")
print("File:", os.path.basename(file_path))
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nFirst five records:")
display(df.head())


# ================================================================
# 4. DATA UNDERSTANDING
# ================================================================

print("\n" + "=" * 75)
print("DATA UNDERSTANDING")
print("=" * 75)

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
for i, column in enumerate(df.columns, start=1):
    print(i, ".", column)

print("\nData Types:")
display(df.dtypes.to_frame("Data Type"))

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all").T)


# ================================================================
# 5. MISSING VALUES AND DUPLICATES
# ================================================================

print("\n" + "=" * 75)
print("MISSING VALUES AND DUPLICATE ANALYSIS")
print("=" * 75)

missing_table = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

print("\nMissing Values:")
display(missing_table)

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nTarget Distribution:")
display(df["Churn"].value_counts().to_frame("Count"))

print("\nTarget Percentage:")
display(
    (
        df["Churn"]
        .value_counts(normalize=True)
        .mul(100)
        .round(2)
        .to_frame("Percentage")
    )
)


# ================================================================
# 6. DATA CLEANING
# ================================================================

print("\n" + "=" * 75)
print("DATA CLEANING")
print("=" * 75)

data = df.copy()

# Remove spaces from column names
data.columns = data.columns.str.strip()

# Remove duplicate rows
before_duplicates = len(data)

data = data.drop_duplicates().reset_index(drop=True)

after_duplicates = len(data)

print(
    "Duplicate rows removed:",
    before_duplicates - after_duplicates
)

# Convert TotalCharges to numeric
data["TotalCharges"] = pd.to_numeric(
    data["TotalCharges"],
    errors="coerce"
)

print(
    "\nMissing TotalCharges after conversion:",
    data["TotalCharges"].isnull().sum()
)

# Fill TotalCharges using median
data["TotalCharges"] = data["TotalCharges"].fillna(
    data["TotalCharges"].median()
)

# Convert target variable
data["Churn"] = data["Churn"].map({
    "Yes": 1,
    "No": 0
})

# Ensure SeniorCitizen remains numeric
data["SeniorCitizen"] = pd.to_numeric(
    data["SeniorCitizen"],
    errors="coerce"
)

print("\nFinal cleaned shape:", data.shape)

print("\nRemaining missing values:")
display(
    data.isnull().sum().to_frame("Missing Values")
)


# ================================================================
# 7. EXPLORATORY DATA ANALYSIS
# ================================================================

print("\n" + "=" * 75)
print("EXPLORATORY DATA ANALYSIS")
print("=" * 75)


# ------------------------------------------------
# 7.1 Churn Distribution
# ------------------------------------------------

plt.figure(figsize=(8, 5))

sns.countplot(
    data=data,
    x="Churn"
)

plt.title("Customer Churn Distribution")
plt.xlabel("Churn (0 = No Churn, 1 = Churn)")
plt.ylabel("Number of Customers")

plt.tight_layout()

plt.savefig(
    os.path.join(FIGURE_DIR, "01_churn_distribution.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------
# 7.2 Churn Percentage
# ------------------------------------------------

churn_rate = data["Churn"].mean() * 100

print(
    f"Overall Customer Churn Rate: {churn_rate:.2f}%"
)

churn_percentage = (
    data["Churn"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)

print("\nNo Churn:", churn_percentage.get(0, 0), "%")
print("Churn:", churn_percentage.get(1, 0), "%")


# ------------------------------------------------
# 7.3 Contract vs Churn
# ------------------------------------------------

plt.figure(figsize=(9, 5))

sns.countplot(
    data=data,
    x="Contract",
    hue="Churn"
)

plt.title("Contract Type vs Customer Churn")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.legend(
    title="Churn",
    labels=["No Churn", "Churn"]
)

plt.tight_layout()

plt.savefig(
    os.path.join(FIGURE_DIR, "02_contract_vs_churn.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------
# 7.4 Internet Service vs Churn
# ------------------------------------------------

plt.figure(figsize=(9, 5))

sns.countplot(
    data=data,
    x="InternetService",
    hue="Churn"
)

plt.title("Internet Service vs Customer Churn")
plt.xlabel("Internet Service")
plt.ylabel("Number of Customers")

plt.tight_layout()

plt.savefig(
    os.path.join(FIGURE_DIR, "03_internet_service_vs_churn.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------
# 7.5 Payment Method vs Churn
# ------------------------------------------------

plt.figure(figsize=(11, 5))

sns.countplot(
    data=data,
    x="PaymentMethod",
    hue="Churn"
)

plt.title("Payment Method vs Customer Churn")
plt.xlabel("Payment Method")
plt.ylabel("Number of Customers")
plt.xticks(rotation=25)

plt.tight_layout()

plt.savefig(
    os.path.join(FIGURE_DIR, "04_payment_method_vs_churn.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------
# 7.6 Tenure vs Churn
# ------------------------------------------------

plt.figure(figsize=(9, 5))

sns.histplot(
    data=data,
    x="tenure",
    hue="Churn",
    kde=True,
    bins=30,
    element="step"
)

plt.title("Customer Tenure Distribution by Churn")
plt.xlabel("Tenure (Months)")
plt.ylabel("Number of Customers")

plt.tight_layout()

plt.savefig(
    os.path.join(FIGURE_DIR, "05_tenure_vs_churn.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------
# 7.7 Monthly Charges vs Churn
# ------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=data,
    x="Churn",
    y="MonthlyCharges"
)

plt.title("Monthly Charges vs Customer Churn")
plt.xlabel("Churn (0 = No Churn, 1 = Churn)")
plt.ylabel("Monthly Charges")

plt.tight_layout()

plt.savefig(
    os.path.join(FIGURE_DIR, "06_monthly_charges_vs_churn.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------
# 7.8 Tenure vs Monthly Charges
# ------------------------------------------------

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=data.sample(
        min(2500, len(data)),
        random_state=42
    ),
    x="tenure",
    y="MonthlyCharges",
    hue="Churn",
    alpha=0.7
)

plt.title("Tenure vs Monthly Charges by Churn")
plt.xlabel("Tenure (Months)")
plt.ylabel("Monthly Charges")

plt.tight_layout()

plt.savefig(
    os.path.join(FIGURE_DIR, "07_tenure_monthlycharges.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------
# 7.9 Correlation Heatmap
# ------------------------------------------------

numeric_eda = data[
    [
        "SeniorCitizen",
        "tenure",
        "MonthlyCharges",
        "TotalCharges",
        "Churn"
    ]
].copy()

plt.figure(figsize=(8, 6))

sns.heatmap(
    numeric_eda.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.savefig(
    os.path.join(FIGURE_DIR, "08_correlation_heatmap.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 8. FEATURE AND TARGET PREPARATION
# ================================================================

print("\n" + "=" * 75)
print("FEATURE AND TARGET PREPARATION")
print("=" * 75)

# Remove target and customer identifier
X = data.drop(
    columns=["Churn", "customerID"],
    errors="ignore"
)

y = data["Churn"]

numeric_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

print("\nNumerical Features:")
print(numeric_features)

print("\nCategorical Features:")
print(categorical_features)

print("\nTotal Numerical Features:", len(numeric_features))
print("Total Categorical Features:", len(categorical_features))


# ================================================================
# 9. PREPROCESSING PIPELINE
# ================================================================

print("\n" + "=" * 75)
print("FEATURE TRANSFORMATION AND PREPROCESSING")
print("=" * 75)

# Make OneHotEncoder compatible with different sklearn versions
try:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            encoder
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            numeric_transformer,
            numeric_features
        ),
        (
            "categorical",
            categorical_transformer,
            categorical_features
        )
    ]
)

print("Preprocessing pipeline created successfully.")


# ================================================================
# 10. TRAIN-TEST SPLIT
# ================================================================

print("\n" + "=" * 75)
print("TRAIN-TEST SPLIT")
print("=" * 75)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Samples:", X_train.shape[0])
print("Testing Samples:", X_test.shape[0])

print("\nTraining Churn Distribution:")
display(
    y_train
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
    .to_frame("Percentage")
)

print("\nTesting Churn Distribution:")
display(
    y_test
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
    .to_frame("Percentage")
)


# ================================================================
# 11. CREATE CLASSIFICATION MODELS
# ================================================================

print("\n" + "=" * 75)
print("CLASSIFICATION MODELS")
print("=" * 75)

models = {

    "Logistic Regression": Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),
            (
                "classifier",
                LogisticRegression(
                    max_iter=2000,
                    random_state=42
                )
            )
        ]
    ),

    "Naive Bayes": Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),
            (
                "classifier",
                GaussianNB()
            )
        ]
    ),

    "Decision Tree": Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),
            (
                "classifier",
                DecisionTreeClassifier(
                    random_state=42
                )
            )
        ]
    ),

    "Support Vector Machine": Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),
            (
                "classifier",
                SVC(
                    probability=True,
                    random_state=42
                )
            )
        ]
    )
}

for model_name in models:
    print("✓", model_name)


# ================================================================
# 12. TRAIN ALL MODELS
# ================================================================

print("\n" + "=" * 75)
print("MODEL TRAINING")
print("=" * 75)

trained_models = {}
predictions = {}
probabilities = {}

for name, model in models.items():

    print(f"\nTraining {name}...")

    model.fit(
        X_train,
        y_train
    )

    y_pred = model.predict(
        X_test
    )

    y_prob = model.predict_proba(
        X_test
    )[:, 1]

    trained_models[name] = model
    predictions[name] = y_pred
    probabilities[name] = y_prob

    print(
        f"✓ {name} training completed."
    )

print("\nAll classification models trained successfully.")


# ================================================================
# 13. MODEL EVALUATION
# ================================================================

print("\n" + "=" * 75)
print("MODEL EVALUATION")
print("=" * 75)

results = []

for name in trained_models:

    y_pred = predictions[name]
    y_prob = probabilities[name]

    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_pred,
        zero_division=0
    )

    auc = roc_auc_score(
        y_test,
        y_prob
    )

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "ROC-AUC": auc
    })

results_df = pd.DataFrame(results)

results_df = (
    results_df
    .sort_values(
        by="F1-Score",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\nModel Performance Comparison:")

display(
    results_df.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-Score": "{:.4f}",
        "ROC-AUC": "{:.4f}"
    })
)

results_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "model_comparison.csv"
    ),
    index=False
)


# ================================================================
# 14. CLASSIFICATION REPORTS
# ================================================================

print("\n" + "=" * 75)
print("CLASSIFICATION REPORTS")
print("=" * 75)

for name in trained_models:

    print("\n" + "-" * 65)
    print(name)
    print("-" * 65)

    print(
        classification_report(
            y_test,
            predictions[name],
            target_names=[
                "Non-Churn",
                "Churn"
            ],
            zero_division=0
        )
    )


# ================================================================
# 15. CONFUSION MATRICES
# ================================================================

print("\n" + "=" * 75)
print("CONFUSION MATRICES")
print("=" * 75)

fig, axes = plt.subplots(
    2,
    2,
    figsize=(13, 10)
)

axes = axes.flatten()

for ax, (name, y_pred) in zip(
    axes,
    predictions.items()
):

    cm = confusion_matrix(
        y_test,
        y_pred
    )

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        ax=ax,
        cbar=False
    )

    ax.set_title(
        f"Confusion Matrix - {name}"
    )

    ax.set_xlabel(
        "Predicted Label"
    )

    ax.set_ylabel(
        "Actual Label"
    )

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIGURE_DIR,
        "09_confusion_matrices.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 16. MODEL COMPARISON GRAPH
# ================================================================

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1-Score",
    "ROC-AUC"
]

comparison_plot_df = (
    results_df
    .set_index("Model")[metrics]
)

ax = comparison_plot_df.plot(
    kind="bar",
    figsize=(13, 7)
)

plt.title(
    "Performance Comparison of Classification Models"
)

plt.xlabel(
    "Machine Learning Model"
)

plt.ylabel(
    "Score"
)

plt.ylim(
    0,
    1
)

plt.xticks(
    rotation=20
)

plt.legend(
    title="Evaluation Metrics",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIGURE_DIR,
        "10_model_comparison.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 17. ROC CURVE COMPARISON
# ================================================================

print("\n" + "=" * 75)
print("ROC CURVE ANALYSIS")
print("=" * 75)

plt.figure(figsize=(10, 7))

for name in trained_models:

    y_prob = probabilities[name]

    fpr, tpr, _ = roc_curve(
        y_test,
        y_prob
    )

    auc_value = roc_auc_score(
        y_test,
        y_prob
    )

    plt.plot(
        fpr,
        tpr,
        linewidth=2,
        label=f"{name} (AUC = {auc_value:.3f})"
    )

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.title(
    "ROC Curve Comparison"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIGURE_DIR,
        "11_roc_curve_comparison.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 18. BEST MODEL SELECTION
# ================================================================

print("\n" + "=" * 75)
print("BEST MODEL SELECTION")
print("=" * 75)

best_model_name = results_df.iloc[0]["Model"]

best_model = trained_models[
    best_model_name
]

best_predictions = predictions[
    best_model_name
]

best_probabilities = probabilities[
    best_model_name
]

best_row = results_df[
    results_df["Model"] == best_model_name
].iloc[0]

print("Selected Best Model:", best_model_name)

print("\nPerformance of Selected Model:")

print(
    f"Accuracy  : {best_row['Accuracy']:.4f}"
)

print(
    f"Precision : {best_row['Precision']:.4f}"
)

print(
    f"Recall    : {best_row['Recall']:.4f}"
)

print(
    f"F1-Score  : {best_row['F1-Score']:.4f}"
)

print(
    f"ROC-AUC   : {best_row['ROC-AUC']:.4f}"
)


# ================================================================
# 19. HYPERPARAMETER EXPERIMENTATION
# ================================================================

print("\n" + "=" * 75)
print("HYPERPARAMETER EXPERIMENTATION")
print("=" * 75)

dt_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "classifier",
            DecisionTreeClassifier(
                random_state=42
            )
        )
    ]
)

param_grid = {
    "classifier__max_depth": [
        3,
        5,
        7,
        10,
        None
    ],

    "classifier__min_samples_split": [
        2,
        5,
        10
    ]
}

grid_search = GridSearchCV(
    estimator=dt_pipeline,
    param_grid=param_grid,
    scoring="f1",
    cv=5,
    n_jobs=-1
)

print("Running Decision Tree hyperparameter search...")

grid_search.fit(
    X_train,
    y_train
)

print("\nBest Parameters:")
print(
    grid_search.best_params_
)

print(
    "\nBest Cross-Validation F1-Score:",
    round(
        grid_search.best_score_,
        4
    )
)

tuned_dt = grid_search.best_estimator_

tuned_pred = tuned_dt.predict(
    X_test
)

tuned_prob = tuned_dt.predict_proba(
    X_test
)[:, 1]

tuned_results = {
    "Accuracy": accuracy_score(
        y_test,
        tuned_pred
    ),

    "Precision": precision_score(
        y_test,
        tuned_pred,
        zero_division=0
    ),

    "Recall": recall_score(
        y_test,
        tuned_pred,
        zero_division=0
    ),

    "F1-Score": f1_score(
        y_test,
        tuned_pred,
        zero_division=0
    ),

    "ROC-AUC": roc_auc_score(
        y_test,
        tuned_prob
    )
}

print("\nTuned Decision Tree Test Performance:")

for metric, value in tuned_results.items():
    print(
        f"{metric}: {value:.4f}"
    )

pd.DataFrame(
    [tuned_results]
).to_csv(
    os.path.join(
        RESULT_DIR,
        "tuned_decision_tree_results.csv"
    ),
    index=False
)


# ================================================================
# 20. K-MEANS CLUSTERING PREPARATION
# ================================================================

print("\n" + "=" * 75)
print("UNSUPERVISED LEARNING - K-MEANS")
print("=" * 75)

# IMPORTANT:
# Churn is deliberately excluded because clustering should
# identify behavioural groups without using the target variable.

cluster_data = data.drop(
    columns=["Churn", "customerID"],
    errors="ignore"
).copy()

cluster_numeric = cluster_data.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

cluster_categorical = cluster_data.select_dtypes(
    include=["object"]
).columns.tolist()

try:
    cluster_encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    cluster_encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )

cluster_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            Pipeline([
                (
                    "imputer",
                    SimpleImputer(
                        strategy="median"
                    )
                ),
                (
                    "scaler",
                    StandardScaler()
                )
            ]),
            cluster_numeric
        ),

        (
            "categorical",
            Pipeline([
                (
                    "imputer",
                    SimpleImputer(
                        strategy="most_frequent"
                    )
                ),
                (
                    "encoder",
                    cluster_encoder
                )
            ]),
            cluster_categorical
        )
    ]
)

cluster_matrix = cluster_preprocessor.fit_transform(
    cluster_data
)

print(
    "Clustering matrix shape:",
    cluster_matrix.shape
)

print(
    "Clustering features:",
    len(cluster_numeric) + len(cluster_categorical)
)


# ================================================================
# 21. ELBOW METHOD
# ================================================================

print("\n" + "=" * 75)
print("ELBOW METHOD")
print("=" * 75)

k_values = range(
    2,
    11
)

inertias = []

for k in k_values:

    kmeans_temp = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans_temp.fit(
        cluster_matrix
    )

    inertias.append(
        kmeans_temp.inertia_
    )

elbow_df = pd.DataFrame({
    "K": list(k_values),
    "Inertia": inertias
})

display(elbow_df)

plt.figure(figsize=(9, 5))

plt.plot(
    list(k_values),
    inertias,
    marker="o",
    linewidth=2
)

plt.title(
    "Elbow Method for Determining Number of Clusters"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Inertia"
)

plt.xticks(
    list(k_values)
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIGURE_DIR,
        "12_elbow_method.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

elbow_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "elbow_results.csv"
    ),
    index=False
)


# ================================================================
# 22. SILHOUETTE SCORE
# ================================================================

print("\n" + "=" * 75)
print("SILHOUETTE SCORE ANALYSIS")
print("=" * 75)

silhouette_results = []

for k in k_values:

    kmeans_temp = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels_temp = kmeans_temp.fit_predict(
        cluster_matrix
    )

    score = silhouette_score(
        cluster_matrix,
        labels_temp
    )

    silhouette_results.append({
        "K": k,
        "Silhouette Score": score
    })

silhouette_df = pd.DataFrame(
    silhouette_results
)

display(
    silhouette_df.style.format({
        "Silhouette Score": "{:.4f}"
    })
)

plt.figure(figsize=(9, 5))

plt.plot(
    silhouette_df["K"],
    silhouette_df["Silhouette Score"],
    marker="o",
    linewidth=2
)

plt.title(
    "Silhouette Score for Different K Values"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Silhouette Score"
)

plt.xticks(
    list(k_values)
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIGURE_DIR,
        "13_silhouette_scores.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

silhouette_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "silhouette_results.csv"
    ),
    index=False
)


# ================================================================
# 23. SELECT OPTIMAL K
# ================================================================

print("\n" + "=" * 75)
print("OPTIMAL NUMBER OF CLUSTERS")
print("=" * 75)

optimal_k = int(
    silhouette_df.loc[
        silhouette_df["Silhouette Score"].idxmax(),
        "K"
    ]
)

best_silhouette = float(
    silhouette_df.loc[
        silhouette_df["Silhouette Score"].idxmax(),
        "Silhouette Score"
    ]
)

print(
    "Selected Optimal K:",
    optimal_k
)

print(
    "Best Silhouette Score:",
    round(
        best_silhouette,
        4
    )
)


# ================================================================
# 24. FINAL K-MEANS MODEL
# ================================================================

print("\n" + "=" * 75)
print("FINAL K-MEANS CLUSTERING")
print("=" * 75)

kmeans_final = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans_final.fit_predict(
    cluster_matrix
)

data["Cluster"] = cluster_labels

print(
    "K-Means clustering completed successfully."
)

cluster_counts = (
    data["Cluster"]
    .value_counts()
    .sort_index()
    .to_frame("Customer Count")
)

display(
    cluster_counts
)

cluster_counts.to_csv(
    os.path.join(
        RESULT_DIR,
        "cluster_counts.csv"
    )
)


# ================================================================
# 25. CLUSTER DISTRIBUTION
# ================================================================

plt.figure(figsize=(9, 5))

sns.countplot(
    data=data,
    x="Cluster"
)

plt.title(
    "Customer Distribution Across Behavioural Segments"
)

plt.xlabel(
    "Customer Cluster"
)

plt.ylabel(
    "Number of Customers"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIGURE_DIR,
        "14_cluster_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 26. CLUSTER PROFILE
# ================================================================

print("\n" + "=" * 75)
print("CUSTOMER CLUSTER PROFILE")
print("=" * 75)

profile_numeric = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges",
    "SeniorCitizen",
    "Churn"
]

cluster_profile = (
    data
    .groupby("Cluster")[profile_numeric]
    .mean()
    .round(2)
)

print(
    "Average characteristics of each customer segment:"
)

display(
    cluster_profile
)

cluster_profile.to_csv(
    os.path.join(
        RESULT_DIR,
        "cluster_profile.csv"
    )
)


# ================================================================
# 27. CATEGORICAL CLUSTER PROFILE
# ================================================================

print("\n" + "=" * 75)
print("CATEGORICAL CHARACTERISTICS OF CLUSTERS")
print("=" * 75)

important_categories = [
    "Contract",
    "InternetService",
    "PaymentMethod"
]

categorical_cluster_tables = {}

for column in important_categories:

    if column in data.columns:

        table = pd.crosstab(
            data["Cluster"],
            data[column],
            normalize="index"
        ).mul(100).round(2)

        categorical_cluster_tables[
            column
        ] = table

        print(
            f"\n{column} Distribution Across Clusters (%):"
        )

        display(table)

        safe_name = (
            column.lower()
            .replace(" ", "_")
        )

        table.to_csv(
            os.path.join(
                RESULT_DIR,
                f"cluster_{safe_name}_distribution.csv"
            )
        )


# ================================================================
# 28. CHURN RATE BY CLUSTER
# ================================================================

print("\n" + "=" * 75)
print("CHURN RATE BY CUSTOMER SEGMENT")
print("=" * 75)

cluster_churn = (
    data
    .groupby("Cluster")["Churn"]
    .agg(
        Customers="count",
        Churned="sum",
        Churn_Rate="mean"
    )
)

cluster_churn["Churn_Rate"] = (
    cluster_churn["Churn_Rate"] * 100
).round(2)

display(
    cluster_churn
)

cluster_churn.to_csv(
    os.path.join(
        RESULT_DIR,
        "cluster_churn.csv"
    )
)


# ================================================================
# 29. CHURN RATE VISUALIZATION BY CLUSTER
# ================================================================

plt.figure(figsize=(9, 5))

sns.barplot(
    data=cluster_churn.reset_index(),
    x="Cluster",
    y="Churn_Rate"
)

plt.title(
    "Churn Rate Across Customer Segments"
)

plt.xlabel(
    "Customer Cluster"
)

plt.ylabel(
    "Churn Rate (%)"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIGURE_DIR,
        "15_cluster_churn_rate.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 30. PCA VISUALIZATION
# ================================================================

print("\n" + "=" * 75)
print("PCA VISUALIZATION")
print("=" * 75)

pca = PCA(
    n_components=2,
    random_state=42
)

pca_data = pca.fit_transform(
    cluster_matrix
)

pca_df = pd.DataFrame(
    pca_data,
    columns=[
        "PC1",
        "PC2"
    ]
)

pca_df["Cluster"] = cluster_labels

pc1_variance = (
    pca.explained_variance_ratio_[0] * 100
)

pc2_variance = (
    pca.explained_variance_ratio_[1] * 100
)

total_variance = (
    pca.explained_variance_ratio_.sum() * 100
)

print(
    f"PC1 Explained Variance: {pc1_variance:.2f}%"
)

print(
    f"PC2 Explained Variance: {pc2_variance:.2f}%"
)

print(
    f"Total Explained Variance: {total_variance:.2f}%"
)

plt.figure(figsize=(11, 7))

sns.scatterplot(
    data=pca_df,
    x="PC1",
    y="PC2",
    hue="Cluster",
    palette="tab10",
    s=60,
    alpha=0.7
)

plt.title(
    "Customer Behavioural Segments Using PCA"
)

plt.xlabel(
    f"Principal Component 1 ({pc1_variance:.1f}% variance)"
)

plt.ylabel(
    f"Principal Component 2 ({pc2_variance:.1f}% variance)"
)

plt.legend(
    title="Cluster"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIGURE_DIR,
        "16_pca_customer_segments.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

pca_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "pca_coordinates.csv"
    ),
    index=False
)


# ================================================================
# 31. CUSTOMER CHURN PROBABILITY FOR ALL CUSTOMERS
# ================================================================

print("\n" + "=" * 75)
print("CUSTOMER RISK ANALYSIS")
print("=" * 75)

# Predict probability for every customer
all_churn_probabilities = (
    best_model
    .predict_proba(X)[:, 1]
)

risk_data = data.copy()

risk_data[
    "Churn_Probability"
] = all_churn_probabilities


# ================================================================
# 32. RISK LEVEL ASSIGNMENT
# ================================================================

def assign_risk(probability):

    if probability < 0.30:
        return "Low Risk"

    elif probability < 0.60:
        return "Medium Risk"

    else:
        return "High Risk"


risk_data["Risk_Level"] = (
    risk_data[
        "Churn_Probability"
    ]
    .apply(assign_risk)
)

print(
    "Risk thresholds used:"
)

print(
    "Low Risk    : Probability < 0.30"
)

print(
    "Medium Risk : 0.30 <= Probability < 0.60"
)

print(
    "High Risk   : Probability >= 0.60"
)

print("\nRisk Distribution:")

risk_distribution = (
    risk_data["Risk_Level"]
    .value_counts()
    .reindex(
        [
            "Low Risk",
            "Medium Risk",
            "High Risk"
        ],
        fill_value=0
    )
    .to_frame("Customers")
)

display(
    risk_distribution
)


# ================================================================
# 33. RISK DISTRIBUTION GRAPH
# ================================================================

plt.figure(figsize=(9, 5))

sns.countplot(
    data=risk_data,
    x="Risk_Level",
    order=[
        "Low Risk",
        "Medium Risk",
        "High Risk"
    ]
)

plt.title(
    "Customer Risk Distribution"
)

plt.xlabel(
    "Risk Level"
)

plt.ylabel(
    "Number of Customers"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIGURE_DIR,
        "17_customer_risk_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 34. RISK DISTRIBUTION BY CLUSTER
# ================================================================

print("\n" + "=" * 75)
print("RISK DISTRIBUTION ACROSS BEHAVIOURAL SEGMENTS")
print("=" * 75)

risk_order = [
    "Low Risk",
    "Medium Risk",
    "High Risk"
]

risk_cluster = pd.crosstab(
    risk_data["Cluster"],
    risk_data["Risk_Level"],
    normalize="index"
).mul(100)

risk_cluster = (
    risk_cluster
    .reindex(
        columns=risk_order,
        fill_value=0
    )
    .round(2)
)

display(
    risk_cluster
)

risk_cluster.to_csv(
    os.path.join(
        RESULT_DIR,
        "risk_distribution_by_cluster.csv"
    )
)

plt.figure(figsize=(10, 6))

risk_cluster.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 6)
)

plt.title(
    "Customer Risk Distribution Across Behavioural Segments"
)

plt.xlabel(
    "Customer Segment"
)

plt.ylabel(
    "Percentage of Customers"
)

plt.legend(
    title="Risk Level",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIGURE_DIR,
        "18_risk_by_cluster.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 35. SEGMENT CHURN RATE ADDED TO RISK DATA
# ================================================================

cluster_churn_lookup = (
    cluster_churn[
        "Churn_Rate"
    ]
    .to_dict()
)

risk_data[
    "Segment_Churn_Rate"
] = (
    risk_data[
        "Cluster"
    ]
    .map(
        cluster_churn_lookup
    )
)


# ================================================================
# 36. HIGH-RISK CUSTOMER IDENTIFICATION
# ================================================================

high_risk_customers = risk_data[
    risk_data["Risk_Level"] == "High Risk"
].copy()

print(
    "Total High-Risk Customers:",
    len(high_risk_customers)
)

print(
    "Percentage of Customers Classified as High Risk:",
    round(
        len(high_risk_customers) /
        len(risk_data) * 100,
        2
    ),
    "%"
)

high_risk_by_cluster = (
    high_risk_customers
    .groupby("Cluster")
    .size()
    .sort_values(
        ascending=False
    )
    .to_frame(
        "High-Risk Customers"
    )
)

print(
    "\nHigh-Risk Customers by Cluster:"
)

display(
    high_risk_by_cluster
)

high_risk_by_cluster.to_csv(
    os.path.join(
        RESULT_DIR,
        "high_risk_by_cluster.csv"
    )
)


# ================================================================
# 37. PRIORITY IDENTIFICATION
# ================================================================

print("\n" + "=" * 75)
print("CUSTOMER PRIORITY IDENTIFICATION")
print("=" * 75)

# Project-level prioritization rule:
#
# High-risk customer + segment churn rate >= 50%
#      -> High Priority
#
# High-risk customer + lower segment churn
#      -> Medium Priority
#
# Medium risk
#      -> Medium Priority
#
# Low risk
#      -> Low Priority

def assign_priority(row):

    if (
        row["Risk_Level"] == "High Risk"
        and row["Segment_Churn_Rate"] >= 50
    ):
        return "High Priority"

    elif row["Risk_Level"] == "High Risk":
        return "Medium Priority"

    elif row["Risk_Level"] == "Medium Risk":
        return "Medium Priority"

    else:
        return "Low Priority"


risk_data[
    "Priority"
] = risk_data.apply(
    assign_priority,
    axis=1
)

priority_distribution = (
    risk_data[
        "Priority"
    ]
    .value_counts()
    .to_frame(
        "Customers"
    )
)

display(
    priority_distribution
)


# ================================================================
# 38. RETENTION RECOMMENDATION ENGINE
# ================================================================

print("\n" + "=" * 75)
print("RETENTION RECOMMENDATION ENGINE")
print("=" * 75)

median_monthly_charge = (
    data["MonthlyCharges"].median()
)


def retention_recommendation(row):

    recommendations = []

    # Risk-based recommendation
    if row["Risk_Level"] == "High Risk":
        recommendations.append(
            "Immediate proactive retention contact"
        )

    elif row["Risk_Level"] == "Medium Risk":
        recommendations.append(
            "Targeted engagement campaign"
        )

    else:
        recommendations.append(
            "Maintain loyalty and engagement"
        )

    # Contract-based recommendation
    if row["Contract"] == "Month-to-month":
        recommendations.append(
            "Offer suitable long-term contract incentives"
        )

    # Tenure-based recommendation
    if row["tenure"] <= 12:
        recommendations.append(
            "Provide onboarding and early-stage support"
        )

    # Monthly charge based
    if row["MonthlyCharges"] >= median_monthly_charge:
        recommendations.append(
            "Provide personalized value-based offers"
        )

    # Internet service based
    if row["InternetService"] == "Fiber optic":
        recommendations.append(
            "Review service quality and customer satisfaction"
        )

    return " | ".join(
        recommendations
    )


risk_data[
    "Retention_Recommendation"
] = risk_data.apply(
    retention_recommendation,
    axis=1
)

print(
    "Sample Retention Recommendations:"
)

display(
    risk_data[
        [
            "customerID",
            "Cluster",
            "Churn_Probability",
            "Risk_Level",
            "Priority",
            "Retention_Recommendation"
        ]
    ]
    .sort_values(
        "Churn_Probability",
        ascending=False
    )
    .head(20)
)


# ================================================================
# 39. SEGMENT-LEVEL RETENTION STRATEGY
# ================================================================

print("\n" + "=" * 75)
print("SEGMENT-LEVEL RETENTION STRATEGIES")
print("=" * 75)

segment_summary = []

for cluster_id in sorted(
    risk_data["Cluster"].unique()
):

    cluster_rows = risk_data[
        risk_data["Cluster"] == cluster_id
    ]

    avg_tenure = (
        cluster_rows["tenure"].mean()
    )

    avg_monthly = (
        cluster_rows["MonthlyCharges"].mean()
    )

    churn_rate_cluster = (
        cluster_rows["Churn"].mean()
        * 100
    )

    avg_probability = (
        cluster_rows[
            "Churn_Probability"
        ].mean()
        * 100
    )

    high_risk_count = (
        cluster_rows[
            "Risk_Level"
        ]
        .eq("High Risk")
        .sum()
    )

    if churn_rate_cluster >= 50:

        strategy = (
            "Immediate retention campaigns, "
            "personalized offers, proactive support "
            "and contract incentives"
        )

    elif churn_rate_cluster >= 25:

        strategy = (
            "Targeted engagement, service improvements, "
            "loyalty benefits and personalized communication"
        )

    else:

        strategy = (
            "Maintain customer engagement, "
            "loyalty programs and relationship-building"
        )

    segment_summary.append({

        "Cluster": cluster_id,

        "Customers": len(
            cluster_rows
        ),

        "Average Tenure": round(
            avg_tenure,
            2
        ),

        "Average Monthly Charges": round(
            avg_monthly,
            2
        ),

        "Churn Rate (%)": round(
            churn_rate_cluster,
            2
        ),

        "Average Churn Probability (%)": round(
            avg_probability,
            2
        ),

        "High-Risk Customers": int(
            high_risk_count
        ),

        "Recommended Strategy": strategy
    })


segment_summary_df = pd.DataFrame(
    segment_summary
)

display(
    segment_summary_df
)

segment_summary_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "segment_summary.csv"
    ),
    index=False
)


# ================================================================
# 40. DECISION TREE FEATURE IMPORTANCE
# ================================================================

print("\n" + "=" * 75)
print("DECISION TREE FEATURE IMPORTANCE")
print("=" * 75)

tree_model = trained_models[
    "Decision Tree"
]

tree_preprocessor = (
    tree_model
    .named_steps[
        "preprocessor"
    ]
)

tree_classifier = (
    tree_model
    .named_steps[
        "classifier"
    ]
)

feature_names = (
    tree_preprocessor
    .get_feature_names_out()
)

importance_df = pd.DataFrame({

    "Feature": feature_names,

    "Importance":
        tree_classifier
        .feature_importances_

})

importance_df = (
    importance_df
    .sort_values(
        "Importance",
        ascending=False
    )
    .head(20)
)

display(
    importance_df
)

importance_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "top_decision_tree_features.csv"
    ),
    index=False
)

plt.figure(figsize=(10, 8))

sns.barplot(
    data=importance_df,
    x="Importance",
    y="Feature"
)

plt.title(
    "Top Features Identified by Decision Tree"
)

plt.xlabel(
    "Feature Importance"
)

plt.ylabel(
    "Feature"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        FIGURE_DIR,
        "19_decision_tree_feature_importance.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 41. FINAL CUSTOMER RISK TABLE
# ================================================================

print("\n" + "=" * 75)
print("TOP HIGH-RISK CUSTOMER TABLE")
print("=" * 75)

risk_columns = [
    "customerID",
    "tenure",
    "Contract",
    "InternetService",
    "MonthlyCharges",
    "TotalCharges",
    "Churn",
    "Cluster",
    "Churn_Probability",
    "Risk_Level",
    "Segment_Churn_Rate",
    "Priority",
    "Retention_Recommendation"
]

final_risk_table = (
    risk_data[
        risk_columns
    ]
    .sort_values(
        by="Churn_Probability",
        ascending=False
    )
    .copy()
)

display(
    final_risk_table.head(25)
)


# ================================================================
# 42. SAVE RISK ANALYSIS
# ================================================================

risk_data_save = risk_data.copy()

risk_data_save[
    "Churn_Probability"
] = (
    risk_data_save[
        "Churn_Probability"
    ] * 100
).round(2)

risk_data_save.to_csv(
    os.path.join(
        RESULT_DIR,
        "customer_risk_analysis.csv"
    ),
    index=False
)

final_risk_table_save = final_risk_table.copy()

final_risk_table_save[
    "Churn_Probability"
] = (
    final_risk_table_save[
        "Churn_Probability"
    ] * 100
).round(2)

final_risk_table_save.head(
    100
).to_csv(
    os.path.join(
        RESULT_DIR,
        "top_100_risk_customers.csv"
    ),
    index=False
)


# ================================================================
# 43. SAVE PROCESSED DATASET
# ================================================================

data.to_csv(
    os.path.join(
        RESULT_DIR,
        "processed_customer_data.csv"
    ),
    index=False
)


# ================================================================
# 44. SAVE MODELS
# ================================================================

print("\n" + "=" * 75)
print("SAVING MACHINE LEARNING MODELS")
print("=" * 75)

joblib.dump(
    best_model,
    os.path.join(
        MODEL_DIR,
        "best_churn_model.pkl"
    )
)

joblib.dump(
    trained_models,
    os.path.join(
        MODEL_DIR,
        "all_classification_models.pkl"
    )
)

joblib.dump(
    cluster_preprocessor,
    os.path.join(
        MODEL_DIR,
        "cluster_preprocessor.pkl"
    )
)

joblib.dump(
    kmeans_final,
    os.path.join(
        MODEL_DIR,
        "kmeans_model.pkl"
    )
)

joblib.dump(
    pca,
    os.path.join(
        MODEL_DIR,
        "pca_model.pkl"
    )
)

print("✓ Best churn model saved.")
print("✓ All classification models saved.")
print("✓ Cluster preprocessing pipeline saved.")
print("✓ K-Means model saved.")
print("✓ PCA model saved.")


# ================================================================
# 45. SAVE PROJECT SUMMARY
# ================================================================

high_risk_count = (
    risk_data[
        "Risk_Level"
    ]
    .eq("High Risk")
    .sum()
)

medium_risk_count = (
    risk_data[
        "Risk_Level"
    ]
    .eq("Medium Risk")
    .sum()
)

low_risk_count = (
    risk_data[
        "Risk_Level"
    ]
    .eq("Low Risk")
    .sum()
)

project_summary = pd.DataFrame([{

    "Total Customers":
        len(data),

    "Total Features":
        X.shape[1],

    "Overall Churn Rate (%)":
        round(
            data["Churn"].mean() * 100,
            2
        ),

    "Best Classification Model":
        best_model_name,

    "Best Accuracy":
        round(
            best_row["Accuracy"],
            4
        ),

    "Best Precision":
        round(
            best_row["Precision"],
            4
        ),

    "Best Recall":
        round(
            best_row["Recall"],
            4
        ),

    "Best F1-Score":
        round(
            best_row["F1-Score"],
            4
        ),

    "Best ROC-AUC":
        round(
            best_row["ROC-AUC"],
            4
        ),

    "Optimal K":
        optimal_k,

    "Silhouette Score":
        round(
            best_silhouette,
            4
        ),

    "Low Risk Customers":
        int(low_risk_count),

    "Medium Risk Customers":
        int(medium_risk_count),

    "High Risk Customers":
        int(high_risk_count)

}])

print("\nPROJECT SUMMARY:")
display(project_summary)

project_summary.to_csv(
    os.path.join(
        RESULT_DIR,
        "project_summary.csv"
    ),
    index=False
)

project_summary.to_json(
    os.path.join(
        RESULT_DIR,
        "project_summary.json"
    ),
    orient="records",
    indent=4
)


# ================================================================
# 46. CREATE MODEL COMPARISON TEXT SUMMARY
# ================================================================

comparison_summary = []

for _, row in results_df.iterrows():

    comparison_summary.append({
        "Model":
            row["Model"],

        "Accuracy":
            round(row["Accuracy"], 4),

        "Precision":
            round(row["Precision"], 4),

        "Recall":
            round(row["Recall"], 4),

        "F1-Score":
            round(row["F1-Score"], 4),

        "ROC-AUC":
            round(row["ROC-AUC"], 4)
    })

comparison_summary_df = pd.DataFrame(
    comparison_summary
)

comparison_summary_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "final_model_comparison.csv"
    ),
    index=False
)


# ================================================================
# 47. FINAL CONSOLE SUMMARY
# ================================================================

print("\n\n")
print("=" * 80)
print("FINAL PROJECT RESULTS")
print("=" * 80)

print("\nDATASET")
print("-" * 80)
print(
    "Total Customers:",
    len(data)
)

print(
    "Total Input Features:",
    X.shape[1]
)

print(
    "Overall Churn Rate:",
    round(
        data["Churn"].mean() * 100,
        2
    ),
    "%"
)


print("\nCLASSIFICATION")
print("-" * 80)

print(
    "Models Evaluated:"
)

for model_name in trained_models:
    print(
        "  ✓",
        model_name
    )

print(
    "\nBest Model:",
    best_model_name
)

print(
    "Accuracy:",
    round(
        best_row["Accuracy"],
        4
    )
)

print(
    "Precision:",
    round(
        best_row["Precision"],
        4
    )
)

print(
    "Recall:",
    round(
        best_row["Recall"],
        4
    )
)

print(
    "F1-Score:",
    round(
        best_row["F1-Score"],
        4
    )
)

print(
    "ROC-AUC:",
    round(
        best_row["ROC-AUC"],
        4
    )
)


print("\nCLUSTERING")
print("-" * 80)

print(
    "Optimal Number of Clusters:",
    optimal_k
)

print(
    "Silhouette Score:",
    round(
        best_silhouette,
        4
    )
)


print("\nRISK ANALYSIS")
print("-" * 80)

print(
    "Low Risk Customers:",
    low_risk_count
)

print(
    "Medium Risk Customers:",
    medium_risk_count
)

print(
    "High Risk Customers:",
    high_risk_count
)


print("\nHIGH-RISK SEGMENTS")
print("-" * 80)

high_risk_cluster_summary = (
    risk_data
    .groupby("Cluster")
    .agg(
        Customers=("Cluster", "size"),
        Average_Churn_Probability=(
            "Churn_Probability",
            "mean"
        ),
        Actual_Churn_Rate=(
            "Churn",
            "mean"
        ),
        High_Risk_Count=(
            "Risk_Level",
            lambda x:
            (x == "High Risk").sum()
        )
    )
)

high_risk_cluster_summary[
    "Average_Churn_Probability"
] *= 100

high_risk_cluster_summary[
    "Actual_Churn_Rate"
] *= 100

high_risk_cluster_summary = (
    high_risk_cluster_summary
    .round(2)
)

display(
    high_risk_cluster_summary
)

high_risk_cluster_summary.to_csv(
    os.path.join(
        RESULT_DIR,
        "high_risk_cluster_summary.csv"
    )
)


# ================================================================
# 48. CREATE COMPLETE PROJECT ZIP
# ================================================================

print("\n" + "=" * 80)
print("CREATING PROJECT OUTPUT ZIP")
print("=" * 80)

zip_base = "/content/customer_churn_project_complete"

zip_file = shutil.make_archive(
    zip_base,
    "zip",
    BASE_DIR
)

print(
    "\nZIP file created successfully:"
)

print(
    zip_file
)


# ================================================================
# 49. DISPLAY GENERATED FILES
# ================================================================

print("\n" + "=" * 80)
print("GENERATED PROJECT FILES")
print("=" * 80)

for root, dirs, files_list in os.walk(BASE_DIR):

    level = root.replace(
        BASE_DIR,
        ""
    ).count(
        os.sep
    )

    indent = "  " * level

    print(
        f"{indent}{os.path.basename(root)}/"
    )

    subindent = "  " * (
        level + 1
    )

    for filename in files_list:

        print(
            f"{subindent}{filename}"
        )


# ================================================================
# 50. FINAL MESSAGE
# ================================================================

print("\n")
print("=" * 80)
print("IMPLEMENTATION COMPLETED SUCCESSFULLY")
print("=" * 80)

print("""
Your project now contains:

✓ Dataset loading and understanding
✓ Data cleaning and preprocessing
✓ Missing-value handling
✓ Duplicate handling
✓ Exploratory Data Analysis
✓ Feature transformation
✓ Train-test split
✓ Logistic Regression
✓ Naive Bayes
✓ Decision Tree
✓ Support Vector Machine
✓ Accuracy
✓ Precision
✓ Recall
✓ F1-Score
✓ ROC-AUC
✓ Classification reports
✓ Confusion matrices
✓ ROC curves
✓ Model comparison
✓ Best model selection
✓ Hyperparameter experimentation
✓ K-Means clustering
✓ Elbow Method
✓ Silhouette Score
✓ Optimal K selection
✓ Cluster profiling
✓ Churn rate by cluster
✓ PCA visualization
✓ Churn probability
✓ Customer risk classification
✓ High-risk customer identification
✓ Priority identification
✓ Retention recommendations
✓ Segment-level retention strategies
✓ Feature importance
✓ Saved ML models
✓ Saved CSV/JSON results
✓ Complete project ZIP

The generated ZIP contains the important files needed later
for your report and potential Google AI Studio web application.
""")

# Automatically download the ZIP
files.download(zip_file)
```
